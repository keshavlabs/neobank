```sql

-- Create a schema for the banking tables
CREATE SCHEMA IF NOT EXISTS neobank.banking;

-- 1. Branches
CREATE TABLE IF NOT EXISTS neobank.banking.branches (
    branch_code STRING NOT NULL,
    branch_name STRING NOT NULL,
    city STRING NOT NULL,
    state STRING NOT NULL,
    region STRING,
    created_at TIMESTAMP DEFAULT CURRENT_TIMESTAMP()
);

-- 2. Customers
CREATE TABLE IF NOT EXISTS neobank.banking.customers (
    customer_id INT NOT NULL,
    first_name STRING NOT NULL,
    last_name STRING NOT NULL,
    date_of_birth DATE NOT NULL,
    pan_number STRING,
    email STRING,
    phone_number STRING,
    kyc_status STRING DEFAULT 'PENDING',
    branch_code STRING NOT NULL,
    created_at TIMESTAMP DEFAULT CURRENT_TIMESTAMP(),
    updated_at TIMESTAMP DEFAULT CURRENT_TIMESTAMP()
);

-- 3. Accounts
CREATE TABLE IF NOT EXISTS neobank.banking.accounts (
    account_id BIGINT NOT NULL,
    customer_id INT NOT NULL,
    account_type STRING NOT NULL,
    balance DECIMAL(18,2) NOT NULL,
    currency STRING DEFAULT 'INR',
    branch_code STRING NOT NULL,
    status STRING DEFAULT 'ACTIVE',
    opened_date DATE NOT NULL,
    created_at TIMESTAMP DEFAULT CURRENT_TIMESTAMP(),
    updated_at TIMESTAMP DEFAULT CURRENT_TIMESTAMP()
);

-- 4. Transactions
CREATE TABLE IF NOT EXISTS neobank.banking.transactions (
    txn_id BIGINT NOT NULL,
    account_id BIGINT NOT NULL,
    txn_type STRING NOT NULL,
    amount DECIMAL(18,2) NOT NULL,
    txn_timestamp TIMESTAMP NOT NULL,
    channel STRING,
    status STRING DEFAULT 'SUCCESS',
    created_at TIMESTAMP DEFAULT CURRENT_TIMESTAMP()
);
```
